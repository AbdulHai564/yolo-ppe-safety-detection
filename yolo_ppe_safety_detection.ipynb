```python
model=YOLO('yolo26n.pt')
```

```python
model.train(data='construction-ppe.yaml',batch=16,epochs=50,imgsz=640)
```

```python
metrics=model.val()
print("mAP50:",metrics.box.map50)
print("mAP50-95:",metrics.box.map)
```

```python
result=model('/content/datasets/construction-ppe/images/val/image151.jpg')
result[0].show()
```
